# How to retrive the bibtex 

EKHI uses a BibTex-compatible data structure to report the publication details. Consequently, the publication details can be easily extracted from a file exported from EKHI.

```python
import json
```

```python
with open('../data/entry_example.json', encoding='utf-8') as f:
    publication_dict = json.load(f)
    publication_details = publication_dict.get('publication_info')
```

This is a Python-dictionary version of the publication details:

```python
publication_details
```

It can be transformed into a BibTex string:

```python
def dict_to_bibtex(entry):
    bibtype = entry.get("document_type", "article")
    bibid = entry.get("ID", "unnamed")

    bib = f"@{bibtype}{{{bibid},\n"
    for key, value in entry.items():
        if key in ["document_type", "ID"]:
            continue
        if value is None:
            continue
        bib += f"  {key} = {{{value}}},\n"
    bib += "}"

    return bib
```

```python
publication_bibtex = dict_to_bibtex(publication_details)
publication_bibtex
```

and exported to a `.bib` file:

```python
with open('../data/publication_example.bib', "w", encoding='utf-8') as f:
    f.write(publication_bibtex)
```
